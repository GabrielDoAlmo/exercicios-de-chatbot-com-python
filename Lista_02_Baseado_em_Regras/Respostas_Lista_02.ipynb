```python
import re
dicionarios = [
    {"padrao" : "ola|oi" , "resposta" : "OLA, COMO VAI?"},
    {"padrao" : "devolver|devoluçao|entregar" , "resposta" : "O prazo de devoluçao dos livros e 4 semanas, poendo ser devolvido antes e cobrado uma taxa extra de 1 real por dia de atraso"},
    {"padrao" : "retirada|pegar|retirar" , "resposta" : "para retirada dos livros e necessario informar numero de telefone e identidade para cadastro"}
]
nome = input("qual o seu nome?\n").lower()
while True:
  botao = input(f"bom dia {nome}\n seja bem vindo a nossa livraria, quais de nossos serviços gostaria de usar?\n 1-ver catalogo\n2-central de ajuda\n3-sair\n")
  if botao == "1":
    opçao1 = input(f"Ola {nome}\nobrigado por visitar nossa biblioteca, atualmente temos as seguintes opçoes de livros\n 1-pequeno principe\n2-menino maluquinho\n3- o senhor dos aneis\n")
    if opçao1 == "1":
      print("voce escolheu o livro pequeno principe\n")
    elif opçao1 == "2":
      print("voce escolheu o livro o menino maluquinho\n")
    elif opçao1 == "3":
      print("voce escolheu o livro o senhor dos aneis\n")
    else:
      print("escolha uma opçao valida(escolha apenas o numero)")
  elif botao == "2":
    botao2 = input("o que deseja na nossa central de ajuda\n posso ajudar com prazos de entrega e termos para a retirada de livros\n").lower()
    for regras in dicionarios:
      if re.search(regras["padrao"], botao2):
        print(regras["resposta"])
        break
    else:
      print("nao entendi, pode digitar novamente?")
  elif botao == "3":
    print(f"fechando o sistema, obrigado pelo acesso {nome}")
    break
  else:
    print("escolha um numero valido")
```
